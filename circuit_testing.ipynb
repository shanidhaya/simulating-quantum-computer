```python
import qutip as qt
from functools import reduce
from circuit_engine import apply_instruction
from quantum_gates import get_state_vector
```

```python
 #Define the Initial State |00>
ket0 = qt.basis(2, 0)
initial_state = qt.tensor(ket0, ket0)
```

```python
# This creates the Bell State: H on qubit 0, then CNOT on control 0, target 1
bell_circuit = [
    ['H', [0]],
    ['CNOT', [0, 1]]
]
```

```python
# Run the entire circuit in ONE line using reduce
# reduce(function, sequence, initial_value)
final_state = reduce(apply_instruction, bell_circuit, initial_state)
```

```python
# View the results
print("Expected Bell State Amplitudes: |00> and |11> should be ~0.707")
print("Our Resulting State Vector:")
print(get_state_vector(final_state).round(3))
```
