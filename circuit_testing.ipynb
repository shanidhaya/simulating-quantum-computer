```python
import qutip as qt
import numpy as np
from functools import reduce
from circuit_engine import apply_instruction, measure_qubit
from quantum_gates import get_state_vector
```

```python
 #Define the Initial State |00>
ket0 = qt.basis(2, 0)
initial_state = qt.tensor(ket0, ket0)
```

```python
# This creates the Bell State: H on qubit 0, then CNOT on control 0, target 1
bell_circuit = [
    ['H', [0]],
    ['CNOT', [0, 1]]
]
```

```python
# Run the entire circuit in ONE line using reduce
# reduce(function, sequence, initial_value)
final_state = reduce(apply_instruction, bell_circuit, initial_state)
```

```python
# View the results
print("Expected Bell State Amplitudes: |00> and |11> should be ~0.707")
print("Our Resulting State Vector:")
print(get_state_vector(final_state).round(3))
```

```python
ket0 = qt.basis(2, 0)
ket1 = qt.basis(2, 1)
initial_state = qt.tensor(ket1, ket0)
```

```python
ket0 = qt.basis(2, 0)
initial_state = ket0
```

```python
# 2. Apply a Hadamard gate to put it in a 50/50 superposition
superposition_state = apply_instruction(initial_state, ['H', [0]])
```

```python
print("--- State BEFORE Measurement ---")
print(np.round(superposition_state.full().flatten(), 3))
print("Notice how both 0 and 1 have a 0.707 amplitude!\\n")
```

```python
# 3. Measure it!
result, collapsed_state = measure_qubit(superposition_state, target_qubit=0)

print(f"--- MEASUREMENT RESULT: {result} ---")
print("State AFTER Measurement (Collapsed):")
print(np.round(collapsed_state.full().flatten(), 3))
```

```python
print("--- Test Case 1: 3-Qubit GHZ State ---")
# 1. Initialize |000>
ket0 = qt.basis(2, 0)
initial_ghz = qt.tensor(ket0, ket0, ket0)

# 2. Circuit to create GHZ state
ghz_circuit = [
    ['H', [0]],
    ['CNOT', [0, 1]], # Entangle Qubit 0 and 1
    ['CNOT', [1, 2]]  # Entangle Qubit 1 and 2
]

# 3. Apply circuit
final_ghz = reduce(apply_instruction, ghz_circuit, initial_ghz)
ghz_vector = get_state_vector(final_ghz).round(3)

print("Expected: Amplitude of ~0.707 at |000> (index 0) and |111> (index 7)")
print("Resulting State Vector:")
print(ghz_vector)
```

```python
print("\n--- Test Case 2: Reversibility / Un-computing ---")
# 1. Initialize to |10>
ket1 = qt.basis(2, 1)
initial_rev = qt.tensor(ket1, ket0)

# 2. Apply gates, then apply them in reverse
rev_circuit = [
    ['H', [0]],
    ['CNOT', [0, 1]],
    # Now reverse it
    ['CNOT', [0, 1]],
    ['H', [0]]
]

final_rev = reduce(apply_instruction, rev_circuit, initial_rev)

print("Initial State (Expected |10>):")
print(get_state_vector(initial_rev).round(3))
print("Final State (Should match Initial exactly):")
print(get_state_vector(final_rev).round(3))
```

```python
print("\n--- Test Case 3: Correlated Measurements ---")
# 1. Create a Bell State |00> + |11>
initial_bell = qt.tensor(ket0, ket0)
bell_circuit = [['H', [0]], ['CNOT', [0, 1]]]
bell_state = reduce(apply_instruction, bell_circuit, initial_bell)

# 2. Measure Qubit 0
res0, collapsed_state_0 = measure_qubit(bell_state, target_qubit=0)
print(f"Measured Qubit 0: Result = {res0}")

# 3. Measure Qubit 1
res1, collapsed_state_1 = measure_qubit(collapsed_state_0, target_qubit=1)
print(f"Measured Qubit 1: Result = {res1}")

# 4. Assert correlation
if res0 == res1:
    print("SUCCESS: Measurements are perfectly correlated! (As expected for a Bell state)")
else:
    print("ERROR: Measurements did not match. Check your collapse logic.")
```
