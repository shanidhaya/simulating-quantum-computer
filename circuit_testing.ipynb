```python
import qutip as qt
import numpy as np
from functools import reduce
from circuit_engine import apply_instruction, measure_qubit
from quantum_gates import get_state_vector
```

```python
 #Define the Initial State |00>
ket0 = qt.basis(2, 0)
initial_state = qt.tensor(ket0, ket0)
```

```python
# This creates the Bell State: H on qubit 0, then CNOT on control 0, target 1
bell_circuit = [
    ['H', [0]],
    ['CNOT', [0, 1]]
]
```

```python
# Run the entire circuit in ONE line using reduce
# reduce(function, sequence, initial_value)
final_state = reduce(apply_instruction, bell_circuit, initial_state)
```

```python
# View the results
print("Expected Bell State Amplitudes: |00> and |11> should be ~0.707")
print("Our Resulting State Vector:")
print(get_state_vector(final_state).round(3))
```

```python
ket0 = qt.basis(2, 0)
ket1 = qt.basis(2, 1)
initial_state = qt.tensor(ket1, ket0)
```

```python
instruction = ['C-H', [0, 1]]
```

```python
final_state = apply_instruction(initial_state, instruction)
```

```python
print("\nExpected output: Qubit 1 should now be a superposition because Qubit 0 was |1>.")
print("Resulting State Vector:")
print(np.round(final_state.full().flatten(), 3))
```

```python
ket0 = qt.basis(2, 0)
initial_state = ket0
```

```python
# 2. Apply a Hadamard gate to put it in a 50/50 superposition
superposition_state = apply_instruction(initial_state, ['H', [0]])
```

```python
print("--- State BEFORE Measurement ---")
print(np.round(superposition_state.full().flatten(), 3))
print("Notice how both 0 and 1 have a 0.707 amplitude!\\n")
```

```python
# 3. Measure it!
result, collapsed_state = measure_qubit(superposition_state, target_qubit=0)

print(f"--- MEASUREMENT RESULT: {result} ---")
print("State AFTER Measurement (Collapsed):")
print(np.round(collapsed_state.full().flatten(), 3))
```
